```python
df.drop_duplicates(inplace=True)
```

```python
df['number_of_cells']=df['number_of_cells'].fillna(df['number_of_cells'].mean())
```

```python
df.dropna(subset=['model','torque_nm','fast_charging_power_kw_dc','fast_charge_port','towing_capacity_kg','cargo_volume_l'],inplace=True)
```

```python
print(df.isnull().sum())
```

```python
import seaborn as sns
import numpy as np
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
```

```python
num_cols=df.select_dtypes(include=['int64','float64']).columns
```

```python
cat_cols=df.select_dtypes(include=['object']).columns
```

```python
import matplotlib.pyplot as plt
df.hist(figsize=(15,15))
plt.show()
```

```python
for col in num_cols:
  plt.figure(figsize=(6,3))
  sns.boxplot(df[col])
  plt.title(col)
  plt.show()
```

```python
corr=df.corr(numeric_only=True)
plt.figure(figsize=(10,6))
sns.heatmap(corr,annot=True,cmap='coolwarm')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.scatterplot(x='battery_capacity_kWh',y='range_km',hue="car_body_type",data=df,s=100,alpha=0.8)
plt.title('Electrical Range vs Battery Capacity by Car Body Type')
plt.xlabel('Battery Capacity (kWh)')
plt.ylabel('Range (km)')
plt.show()
```

```python
for col in num_cols:
  Q1=df[col].quantile(0.25)
  Q2=df[col].quantile(0.75)
  IQR=Q2-Q1
  lower_bound=Q1-1.5*IQR
  upper_bound=Q2+1.5*IQR
  df[col]=np.where(df[col]>upper_bound,upper_bound,df[col])
  df[col]=np.where(df[col]<lower_bound,lower_bound,df[col])
```

```python
for col in num_cols:
  plt.figure(figsize=(6,8))
  plt.boxplot(df[col])
  plt.title(col)
  plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x='seats',data=df)
plt.title("number of seats distribution")
plt.show()
```

```python
sns.pairplot(df)
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.countplot(x='car_body_type',data=df,palette='pastel')
plt.title("Car Body Type Distribution")
plt.show()
```
